```python
import pandas as pd
```

```python
from sklearn.naive_bayes import MultinomialNB
```

```python
import os
```

```python
def read_data(root_dir):
    corpus = []
    classes = []
    for class_name in os.listdir(root_dir):
        class_dir = os.path.join(root_dir,class_name)
        for file_name in os.listdir(class_dir):
            file_path = os.path.join(class_dir,file_name)
            with open(file_path,'r') as f:
                word_counts = {}
                for line in f:
                    word, count = line.split()
                    count = int(count)
                    word_counts[word] = count
                corpus.append(word_counts)
                classes.append(class_name)

    return corpus, classes
```

```python
X_train, y_train = read_data('./data/Ebart/VektoriEbart-5/Trening')
```

```python
len(X_train)
```

```python
X_train[0]
```

```python
from sklearn.feature_extraction import DictVectorizer
```

```python
vectorizer = DictVectorizer()
```

```python
vectorizer.fit(X_train)
```

```python
X_train = vectorizer.transform(X_train)
```

```python
X_train.toarray()
```

```python
X_train = pd.DataFrame(X_train.toarray(), columns=vectorizer.feature_names_)
```

```python
X_train
```

```python
model = MultinomialNB()
```

```python
model.fit(X_train,y_train)
```

```python
y_train_pred = model.predict(X_train)
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_score(y_train,y_train_pred)
```

```python
X_test, y_test = read_data('./data/Ebart/VektoriEbart-5/Testing')
```

```python
X_test = vectorizer.transform(X_test)
```

```python
X_test = pd.DataFrame(X_test.toarray(), columns=vectorizer.feature_names_)
```

```python
y_test_pred = model.predict(X_test)
```

```python
accuracy_score(y_test,y_test_pred)
```

```python
from sklearn.metrics import confusion_matrix
```

```python
confusion_matrix(y_test,y_test_pred)
```
