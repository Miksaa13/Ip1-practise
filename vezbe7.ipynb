```python
df = pd.read_csv('data/balloons.csv') #nisam uspeo da nadjem ballons skup -.-
```

```python
X = df[df.columns[:-1]]
```

```python
print(X.describe())
```

```python
y = df.iloc[:, -1]
y.shape
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=1/3, random_state=13, stratify=y)
```

```python
X_train.shape
```

```python
y_train.shape
```

```python
X_test.shape
```

```python
from sklearn.naive_bayes import CategoricalNB
```

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
oe = OrdinalEncoder()
```

```python
oe.fit(X_train)
```

```python
oe.categories_
```

```python
X_train = oe.transform(X_train)
```

```python
X_test = oe.transform(X_test)
```

```python
model = CategoricalNB()
```

```python
model.fit(X_train, y_train)
```

```python
y_train_predict = model.predict(X_train)
```

```python
from sklearn.metrics import confusion_matrix
```

```python
confusion_matrix(y_train, y_train_predict)
```

```python
y_test_predicted = model.predict(X_test)
```

```python
confusion_matrix(y_test, y_test_predicted)
```

```python
from sklearn.pipeline import Pipeline
```

```python
pipe = Pipeline([('ordinal encoder', OrdinalEncoder()), ('classifier', CategoricalNB())])
```

```python
pipe.fit(X_train,y_train)
```
